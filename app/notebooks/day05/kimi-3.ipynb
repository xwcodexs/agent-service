```python
from langchain_mcp_adapters.client import MultiServerMCPClient
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model
# ===================Step 1.初始化模型=====================
# 1.加载环境变量
from dotenv import load_dotenv
from langchain_core.messages import HumanMessage, AIMessage

load_dotenv()

import os
base_url = os.getenv("DASHSCOPE_BASE_URL")
api_key = os.getenv("DASHSCOPE_API_KEY")

# 2.初始化模型
model = init_chat_model(
    model="kimi-k3", # 模型名称，这里可以自定义，我们用的是阿里的qwen-max
    model_provider="openai", # 如果是Langchain不支持的模型，用openai的规范
    base_url=base_url,
    api_key=api_key
)
from langchain.tools import tool
@tool("square_root",description="计算一个数的平方根")
def square_root(x: float)->float:
    return x ** 0.5

client = MultiServerMCPClient(
    {
        "time-mcp": {
            "transport": "stdio",
            "args": [
                "-y",
                "time-mcp"
            ],
            "command": "npx"
        }
    }
)
# 3.2.用client拉取tool
tools = await client.get_tools()
agent = create_agent(model=model, tools=tools, system_prompt="像海盗一样说话")
response = await agent.ainvoke({"messages":[HumanMessage(content="几点了？")]})
```
